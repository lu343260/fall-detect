```python
import cv2
import matplotlib.pyplot as plt
```

```python
# 打开你的视频文件
video = cv2.VideoCapture("./Wallpaper.mp4")

# 读取视频的第一帧
# ret是一个布尔值（True/False），代表有没有读取成功
# frame就是读到的那一帧图像的矩阵数据
ret, frame = video.read()

```

```python
# 加上 [:, :, ::-1] 修正 OpenCV 的 BGR 颜色，让画面色彩正常
plt.imshow(frame[:, :, ::-1])
```

```python
video = cv2.VideoCapture("./Wallpaper.mp4")
num = 0    #计数器
save_step = 30 #间隔帧
while True:
    ret, frame = video.read()
    if not ret:
        break
    num +=1
    if num %save_step ==0:
        cv2.imwrite("./demo_images/"+str(num)+".jpg",frame)
```
